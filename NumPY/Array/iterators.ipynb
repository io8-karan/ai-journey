arr = np.arange(1, 25).reshape(4, 6)

Iterate through it and print only elements satisfying:

element is divisible by 3
AND
element > 10

Do not use Boolean indexing.

```python
import numpy as np
arr=np.arange(1,25).reshape(4,6)
for row in arr:
    for col in row:
      if col > 10 and col % 3 ==0 :
        print(col)
```

14. nditer() — 2 Questions
Q27 — Intermediate

Create:

arr = np.arange(1, 25).reshape(4, 6)

Using nditer():

print every element
print only even elements
calculate their sum

Do not use normal nested for loops.



```python
import numpy as np

arr = np.arange(1, 25).reshape(4, 6)

total = 0

for element in np.nditer(arr):
    print(element)

    if element % 2 == 0:
        print("Even:", element)
        total += element

print("Sum of even elements:", total)
```

Q28 — Advanced

Create:

arr = np.arange(1, 25).reshape(4, 6)

Using nditer():

Iterate over all elements.
Identify elements divisible by 3.
Replace those values with -1.
Print the final array.

Your solution must use the appropriate nditer() configuration for modifying the underlying array.

```python
arr=np.arange(1,25).reshape(4,6)
for x in np.nditer(arr,op_flags=["readwrite"]):
    if x % 3 == 0:
        x[...]=-1
print(arr)
```

Q30 — Advanced

Create:

arr = np.arange(1, 49).reshape(2, 3, 8)

Using ndenumerate():

Find the maximum value. 

Store its complete index.

Find the minimum value.

Store its complete index.

Print both index positions.


Do not use np.argmax() or np.argmin().


```python
arr = np.arange(1, 49).reshape(2, 3, 8)
max=0
min=0
for index,value in np.ndenumerate(arr):
    if value > max:
        max=value
        max_index=index
    if value < min:
        min=value
        min_index=index
print(f"MAximum value {max} at index {max_index}")
print(f"Minimum Value {min} at {min_index}")
```
