```python
import numpy as np

a = np.array([
    [[1,2],
     [3,4]],

    [[5,6],
     [7,8]]
            ])

b = np.array([
    [[11,12],
     [13,14]],

    [[15,16],
     [17,18]]
])
print(a.shape)   # (2,2,2)
print(b.shape)   # (2,2,2)
result=np.concatenate((a,b),axis=2)
print(result)
print(result.shape)
```

# stack 1d-2d

```python
import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])
print(a.shape)   # (2,2,2)
print(b.shape)   # (2,2,2)
result=np.stack((a,b),axis=1)
print(result)
print(result.shape)
```

```python
a=np.array([[1,2,3],[4,5,6],[1,2,3],[4,5,6]])
b=np.array([[11,22,43],[54,56,67],[1,2,3],[4,5,6]])
c=np.array([[11,22,43],[54,56,67],[1,2,3],[4,5,6]])
print(a.shape)   # (2,2,2)
print(b.shape)   # (2,2,2)
result=np.stack((a,b,c),axis=2)
print(result)
print(result.shape)
```
