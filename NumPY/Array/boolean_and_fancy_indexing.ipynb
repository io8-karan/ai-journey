Q2

Print only those values that are:

greater than 10
less than 20
even
Q3

Create a 4×5 random integer array (1–100).

Replace all numbers less than 50 with 0 using Boolean Indexing.

```python
import numpy as np
```

```python
arr=np.arange(1,37,dtype=int)
new=arr.reshape(6,6)
print(new)
print("\n")
new[new % 4==0]=-1
print(new)
```

```python
result=new[(new>10)& (new<20) & (new % 2 == 0)]
print(result)
```

```python
import numpy as np

arr=np.random.randint(1,101,size=(4,5))
print(arr)
print("\n")

arr[arr< 50]=0
print(arr)
```

Q3

Create a 6×6 array.

Extract rows

1

and columns

0

```python
arr=np.random.randint(1,50,size=(6,6))
print(arr)
print("\n")
new=arr[[0,2,5]]
print(new)
```

```python
arr=np.random.randint(1,50,size=(5,5))
print(arr)
print("\n")
new=arr[[0,2,4],[4,2,1]]
print("Result in 1D Array=",new)
```

```python
arr=np.random.randint(1,50,size=(6,6))
print(arr)
print("\n")
new=arr[[1,3,5]]
new1=new[:,[0,2,4]]
print(new)
print("\n")
print(new1)
```
