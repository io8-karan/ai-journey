```python
import numpy as np

# 1. Create a 1D array
arr1 = np.array([10, 20, 30])

# 2. Create a 2D array
arr2 = np.array([[1, 2], [3, 4]])

# 3. Create a 3D array
arr3 = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

# Print their types
print(arr1)
print("\n")
print(arr2)
print("\n")
print(arr3)
```

```python
import numpy as np

# Using ndmin
a = np.array([1, 2, 3], ndmin=4)
print(a.shape)      # (1, 3)

# Using reshape
b = np.array([1, 2, 3]).reshape(3, 1)
print(b.shape)      # (3, 1)
```
