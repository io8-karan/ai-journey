1. reshape() — 2 Questions
Q1 — Intermediate

Create a NumPy array containing numbers 1–48 with shape (4, 12).

Using only reshape() and indexing/slicing, transform it into a (3, 4, 4) array such that the elements remain in their original C-order.

Then:

Extract the second 3D block.
From that block, extract the last 2 rows.
Print its shape.


```python
import numpy as np

arr = np.arange(1,49,dtype=int).reshape(4,12)

new = arr.reshape(3,4,4)

print(arr)
print("\n")

print(new)

print("Second 3D Block:\n", new[1])

print("Last 2 Rows from Second 3D Block:\n", new[1, -2:])

print("Shape:", new[1, -2:].shape)
```

Q10 — Advanced

Create:

arr = np.arange(1, 49).reshape(2, 3, 8)

Use transpose() to perform:

(2,3,8)
→
(8,2,3)
→
(3,8,2)
→
(2,3,8)

Without creating a new array using np.array().

Verify that the final arrangement is equivalent to the original.

```python
arr = np.arange(1, 49).reshape(2, 3, 8)

print(arr.shape)

new = arr.transpose(2, 0, 1)
print(new.shape)
# (8, 2, 3)

new1 = new.transpose(2,0,1)
print(new1.shape)
# (3, 8, 2)

new2 = new1.transpose(2, 0, 1)
print(new2.shape)
# (2, 3, 8)
print("array at position (8,2,3)\n",new)
print("\n")
print("array at position (3,8,2)\n",new1)
print("\n")
print("array at position (2,3,8)\n",new2)
```

Use expand_dims() to generate all of these shapes:

(1, 3, 4)
